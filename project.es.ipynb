#### Ejercicio 01. Análisis de datos básico con estructuras nativas de python.
Crea una lista con las calificaciones de 5 estudiantes: [8.5, 9.2, 7.8, 8.9, 10].

- Calcula el promedio de las calificaciones.

```python
calificaciones= [8.5, 9.2, 7.8, 8.9, 10]
promedio= sum(calificaciones) / len(calificaciones)
print(promedio)

# otra opción sería:
total= 0
for notas in calificaciones:
    total += notas
    
promedio2= total / len(calificaciones)
print(promedio2)
```

- Encuentra la calificación más alta y la más baja.

```python
maximo= max(calificaciones)
minimo= min(calificaciones)

print("la calificación más alta es:", maximo)
print("la calificación más baja es:", minimo)

#o puede ser directamente imprimiendo:
# calificaciones= [8.5, 9.2, 7.8, 8.9, 10]
#print(max(calificaciones))
#print(min(calificaciones))
```

#### Ejercicio 01. ¿Cuál es la casa más cara en todo el dataset? (★☆☆)

Imprime la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle del Prado, Nº20 es la más cara y su precio es de 5000000 USD`

```python
print(dataset.columns) # --> sólo para poder ver los nombres de las columnas

casa_cara= dataset[dataset['price']== max(dataset['price'])]
print(f"La casa con dirección en {str(casa_cara.address.iloc[0])} es la casa más cara y su precio es de {casa_cara.price.iloc[0]} USD.")

casa_mas_cara= dataset.loc[dataset['price'].idxmax()]
print(f"La casa con dirección en {casa_mas_cara['address']} es la casa más cara y su precio es de {casa_mas_cara['price']} USD.")
```

#### Ejercicio 02. ¿Cuál es la casa más barata del dataset? (★☆☆)

Este ejercicio es similar al ejercicio anterior, solo que ahora buscamos la casa con el precio más bajo. Recuerda imprimir la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle Alcalá, Nº58 es la más barata y su precio es de 12000 USD`

```python
casa_valor_mayor_0= dataset[dataset['price']>0]
casa_barata= casa_valor_mayor_0[casa_valor_mayor_0['price']== min(casa_valor_mayor_0['price'])]
print(f"La casa con dirección en {str(casa_barata.address.values[0])} es la casa más cara y su precio es de {casa_barata.price.iloc[0]} USD.")
```
